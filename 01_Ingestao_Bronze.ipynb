```python
# 1. Credenciais do Azure Data Lake
storage_account_name = "datalakeecommerce123"
storage_account_access_key = "SUA_CHAVE_AQUI_NO_DATABRICKS_SECRETS"

spark.conf.set(
    f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net",
    storage_account_access_key
)

# 2. Caminhos dos contêineres
path_raw = f"abfss://raw@{storage_account_name}.dfs.core.windows.net/*.csv"
path_bronze = f"abfss://bronze@{storage_account_name}.dfs.core.windows.net/ecommerce"

# 3. Leitura do arquivo CSV bruto
df_raw = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(path_raw)
)

# 4. Escrita na Camada Bronze (Formato Delta)
(df_raw.write
    .format("delta")
    .mode("overwrite")
    .save(path_bronze)
)

print("Dados salvos na Camada Bronze com sucesso!")
```
